```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window
```

```python
pn = spark.table("workspace.silver.crm_products")
pc = spark.table("workspace.silver.erp_product_category")

df = pn.join(pc, pn['category_key'] == pc["category_id"], "left")\
    .select(
        row_number().over(Window.orderBy(pn['product_start_date'], pn['product_id'])).alias('product_key'),
        pn['product_id'], pn['product_key'].alias("source_product_key"), pn["product_name"],
        pn['category_key'], pc['category'], pc['subcategory'], pc['maintenance_flag'],
        pn['product_line'], pn['product_start_date']
    )
```
