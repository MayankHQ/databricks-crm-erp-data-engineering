```python
query = """
SELECT 
    sd.order_number,
    pr.source_product_key as product_key,
    cu.customer_key,
    sd.order_date,
    sd.ship_date,
    sd.due_date,
    sd.sales_amount,
    sd.quantity,
    sd.price
FROM silver.crm_sales sd
LEFT JOIN gold.dim_products pr ON sd.product_number = pr.source_product_key
LEFT JOIN gold.dim_customers cu ON sd.customer_id = cu.customer_id 
"""

df = spark.sql(query)
```

```python
display(df.limit(5))
```

```python
df.write.mode("overwrite").saveAsTable('workspace.gold.fact_sales')
```
